```python
from fabricops_kit import finalize_pipeline, orchestrate_read, orchestrate_write, setup_notebook, setup_pipeline


context = setup_notebook()

# For shared config, use the workspace hosting 00_env_config:
# context = setup_notebook(
#     config_workspace_id="<configuration-workspace-id>",
# )
```

# 1. Setup Pipeline

**Ungoverned** to run without Data Contracts,

**Dry Run** to preview without writing, 

**Governed** to enforce activated Data Contracts. Pipeline source participation and reporting are automatic.

| Capability | Dry Run | Ungoverned | Governed |
|---|---|---|---|
| Automatic profiling | Yes | Yes | Yes |
| Publish data | No | Yes | Yes |
| Persist observations | No | Yes | Yes |
| Activated Data Contract required | No | No | Yes |
| Enforce contract guardrails | Yes | No | Yes |


```python
# Ungoverned automatically registers or recovers this notebook's permanent UUID.
# Retain an explicit UUID in source only for cross-workspace promotion.
PIPELINE_ID = None
PIPELINE_MODE = "Ungoverned"  # Dry Run, Ungoverned, Governed

pipeline = setup_pipeline(
    pipeline_id=PIPELINE_ID,
    pipeline_mode=PIPELINE_MODE,
)
PIPELINE_ID = pipeline.pipeline_id
```

# 2. Read

Configure each source directly in its `orchestrate_read()` call. Add another Read block when the transformation needs another governed source.

Read and Write expose Data Quality diagnostics. Grain discovery and Frequency profiling follow the active Data Contract profiling policy, independently of Guardrail enforcement. Mode-specific `read_parameters`, `target_table_id`, and `write_parameters` are demonstrated in Steps 2B and 2C.


```python
# Keep complete Read results for transformation and lineage.
sources = {}
```

## READ 1 — Example source


```python
source = orchestrate_read(

    store="Bronze",
    schema="demo",
    table_name="source_table",  # Replace with the governed source table.
    read_mode="full",
    query=None,

    verbose=True,
    dq_diagnostics="sample",  # "none" keeps metrics only; "full" returns failed rows.
)
sources["source"] = source

# Inspect the actual rows returned by this Read.
display(sources["source"]["dataframe"].limit(10))
```

# 3. Transform

Keep project-owned PySpark logic in this cell. The identity selection below is intentionally safe to replace with joins, filters, derived columns, or aggregations.


```python
transformed_df = sources["source"]["dataframe"].select("*")

# Optional development inspection.
# display(transformed_df)
```

# 4. Write

Configure the target directly in `orchestrate_write()`. FabricOps associates successful Reads with the target as pipeline participants; these associations do not prove DataFrame lineage.

For managed Sensitive Data token mappings, FabricOps uses the configured Governance store. Only supply `existing_mapping` when handling mappings yourself; see the [`orchestrate_write()` reference](https://voycepeh.github.io/FabricOps-Starter-Kit/api/reference/orchestrate_write/).


```python
# Keep complete Write results for optional inspection.
writes = {}
```

## WRITE 1 — Example target


```python
writes["target"] = orchestrate_write(
    transformed_df,

    store="Silver",
    schema="demo",
    table_name="target_table",  # Replace with the governed target table.
    write_mode="overwrite",

    repartition_by=None,

    verbose=True,
    dq_diagnostics="sample",  # "none" keeps metrics only; "full" returns failed rows.
)

# Inspect the DataFrame submitted to this Write.
display(writes["target"]["dataframe"].limit(10))

# Optional: inspect Data Quality diagnostics and caller-managed token mapping.
# display(writes["target"].get("dq_result"))
# display((writes["target"].get("sensitive_result") or {}).get("support_mapping"))

# Optional: read the persisted target after a successful persistent execution.
# from fabricops_kit import read_lakehouse_table
# display(read_lakehouse_table(store="Silver", schema="demo", table_name="target_table").limit(10))
```

Complete the run after every Read, transformation and Write. Dry Run keeps its summary in the session only. Ungoverned and Governed record the real outcome; Governed also releases its activation reservation. A stopped notebook remains incomplete.
