# 03_governance — formal governance review template

Select profiled data, author governance enrichment, author guardrail rules, then complete formal governance review.

02_pipeline may create draft, pending, or active-pending-review records during engineering work. Formal approval, rejection, replacement, and deactivation happen only in 03_governance.


## 1. Run `00_env_config`


```python
%run 00_env_config
```

## 2. Import governance widgets


```python
from fabricops_kit import (
    widget_select_guardrail_target,
    widget_enrich_table_metadata,
    widget_author_guardrail_rules,
    widget_review_table_governance,
)
```

## 3. Select profiled data

Select a profiled source table or pipeline output from the metadata catalogue. Run 02_pipeline profiling first if the table or output is not available for selection.


```python
guardrail_state = widget_select_guardrail_target(spark_session=spark)
```

## 4. Author enrichment

Review and update column context, classification, and other enrichment details for the selected table.

Engineering users can draft enrichment in 02_pipeline. Governance users can author or refine enrichment here before formal review.


```python
enrichment_widget = widget_enrich_table_metadata(
    guardrail_state,
    spark_session=spark,
    source_notebook_type="03_governance",
    created_by_role="governance",
)
```

## 5. Author guardrail rules

Create or update schema, freshness, profile, or data-quality guardrail rules for the selected table.

Rules authored here are governance-owned. Rules drafted in 02_pipeline can be reviewed and finalised here.


```python
guardrail_authoring_widget = widget_author_guardrail_rules(
    guardrail_state,
    spark_session=spark,
    source_notebook_type="03_governance",
    created_by_role="governance",
)
```

## 6. Review governance records

Review enrichment and guardrail records for the selected table.

Approve, reject, replace, deactivate, or inspect history for records that require a formal governance decision.


```python
governance_review_widget = widget_review_table_governance(
    guardrail_state,
    spark_session=spark,
)
```
