# 03_governance — formal governance review

`03_governance` is the formal review notebook. FabricOps separates operational activation from formal governance review. Target selection, enrichment authoring, guardrail authoring, and governance review are separate runnable steps. Formal review happens only in `03_governance`; `02_pipeline` may create draft, pending, or active pending review records, but it cannot formally approve or reject records.


## 1. Run 00_env_config

Load environment configuration and metadata routing before any governance widget runs.


```python
%run 00_env_config
```

## 2. Import governance widgets

Import each widget explicitly so governance users can run the step they need.


```python
from fabricops_kit import (
    widget_select_guardrail_target,
    widget_enrich_table_metadata,
    widget_author_guardrail_rules,
    widget_review_table_governance,
)

ENV_NAME = ENV
```

## 3. Select profiled table or target

Select a table from `METADATA_DATA_CATALOGUE`. Inspect the selected target before moving to authoring or review.


```python
guardrail_state = widget_select_guardrail_target(
    CONFIG,
    ENV_NAME,
    spark_session=spark,
)
```

## 4. Author enrichment

Governance users may author enrichment records here. Authoring can save drafts, submit for review, or apply now; formal review decisions remain in the review step.


```python
enrichment_widget = widget_enrich_table_metadata(
    guardrail_state,
    config=CONFIG,
    env=ENV_NAME,
    spark_session=spark,
)
```

## 5. Author guardrail rules

Author schema, freshness, profile behavior, and DQ guardrail records independently from formal review.


```python
guardrail_authoring_widget = widget_author_guardrail_rules(
    guardrail_state,
    config=CONFIG,
    env=ENV_NAME,
    spark_session=spark,
)
```

## 6. Review table governance

Formal review happens only in `03_governance`. Governance users can approve, approve and activate, reject, replace, deactivate, and view history. Replacing a record appends a new version and marks the previous version as superseded.


```python
governance_review_widget = widget_review_table_governance(
    guardrail_state,
    config=CONFIG,
    env=ENV_NAME,
    spark_session=spark,
)
```
